```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
software_list = []
hardware_list = []
erro_list = []


for i in range(3, 8):
    df = pd.read_csv(f'../resultados_software/eta0/et{i}/resultados_et{i}_eta0.csv')
    software_value = df['Saída_Linear'].to_numpy()
    hardware_value = np.loadtxt(f'../mem/central_barrel/eta0/et{i}/et{i}_eta0_output_linear.txt')
    erro_value = np.abs(software_value - hardware_value)

    software_list.append(software_value)
    hardware_list.append(hardware_value)
    erro_list.append(erro_value)

for i in range(3, 8):
    df = pd.read_csv(f'../resultados_software/eta1/et{i}/resultados_et{i}_eta1.csv')
    software_value = df['Saída_Linear'].to_numpy()
    hardware_value = np.loadtxt(f'../mem/central_barrel/eta1/et{i}/et{i}_eta1_output_linear.txt')
    erro_value = np.abs(software_value - hardware_value)

    software_list.append(software_value)
    hardware_list.append(hardware_value)
    erro_list.append(erro_value)
    

software_2d_np_array = np.array(software_list)
hardware_2d_np_array = np.array(hardware_list)
erro_2d_np_array = np.array(erro_list)

software_np_array = software_2d_np_array.flatten()
hardware_np_array = hardware_2d_np_array.flatten()
erro_np_array = erro_2d_np_array.flatten()

print(f'Erro shape: {erro_2d_np_array.shape}')
print(f'Software shape: {software_2d_np_array.shape}')
print(f'Hardware shape: {hardware_2d_np_array.shape}')

print(f'Erro 1D shape: {erro_np_array.shape}')
print(f'Software 1D shape: {software_np_array.shape}')
print(f'Hardware 1D shape: {hardware_np_array.shape}')
```

```python
# ----- Gráfico de dispersão -----


plt.figure()
# plt.subplot(1, 2, 1)
plt.scatter(software_np_array, hardware_np_array, color='blue')
plt.plot(software_np_array, software_np_array, color='red', linestyle='--', label='Ideal (y = x)')
plt.xlabel('Saída - Software (ponto flutuante)')
plt.ylabel('Saída - Hardware (ponto fixo)')
plt.title('Dispersão: Software vs Hardware')
plt.legend()
plt.grid(True)

# ----- Histograma do erro absoluto -----
plt.figure()
# plt.subplot(1, 2, 2)
plt.hist(erro_np_array, bins=20, color='green', edgecolor='black')
plt.xlabel('Erro absoluto |software - hardware|')
plt.ylabel('Frequência')
plt.title('Histograma do Erro Absoluto')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Calculate RMSE and standard deviation for each pair
rmses = []
stds = []

for s, h in zip(software_list, hardware_list):
    error = s - h
    rmse = np.sqrt(np.mean(error**2))
    std = np.std(error)
    rmses.append(rmse)
    stds.append(std)

labels = [f'Pair {i+1}' for i in range(10)]

# Plot RMSE
plt.figure(figsize=(8, 4))
plt.bar(labels, rmses, color='skyblue')
plt.ylabel('RMSE')
plt.title('RMSE for Each Software/Hardware Pair')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()

# Plot Standard Deviation
plt.figure(figsize=(8, 4))
plt.bar(labels, stds, color='orange')
plt.ylabel('Standard Deviation')
plt.title('Standard Deviation of Error for Each Pair')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
labels = [f'Pair {i+1}' for i in range(10)]
x = np.arange(len(labels))

plt.figure(figsize=(8, 4))
plt.errorbar(x, rmses, yerr=stds, fmt='o', capsize=5, color='blue', ecolor='red')
plt.xticks(x, labels, rotation=45)
plt.ylabel('RMSE')
plt.title('RMSE com Barra de Erro (Desvio Padrão) para Cada Par')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
# Normalize RMSE between -1 and 1
rmses_norm = 2 * (np.array(rmses) - np.min(rmses)) / (np.max(rmses) - np.min(rmses)) - 1


labels = [f'Pair {i+1}' for i in range(10)]
x = np.arange(len(labels))

plt.figure(figsize=(8, 4))
plt.errorbar(x, rmses_norm, yerr=stds, fmt='o', capsize=5, color='blue', ecolor='red')
plt.xticks(x, labels, rotation=45)
plt.ylabel('Normalized RMSE')
plt.title('Normalized RMSE with Normalized Error Bars (-1 to 1) for Each Pair')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
