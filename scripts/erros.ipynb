```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Dados de saída do software (ponto flutuante) e hardware (ponto fixo)
software = np.array([
    -8.455830, 3.604886, -8.735584, 3.464222, -9.816278,
    3.317434, -7.079494, 4.024972, 5.117783, -6.774280
])

hardware = np.array([
    -8.456940, 3.605286, -8.734772, 3.464478, -9.815704,
    3.317993, -7.078400, 4.024536, 5.117706, -6.774719
])

# Erro absoluto
erro_absoluto = np.abs(software - hardware)
```

```python
# ----- Gráfico de dispersão -----
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(software, hardware, color='blue')
plt.plot(software, software, color='red', linestyle='--', label='Ideal (y = x)')
plt.xlabel('Saída - Software (ponto flutuante)')
plt.ylabel('Saída - Hardware (ponto fixo)')
plt.title('Dispersão: Software vs Hardware')
plt.legend()
plt.grid(True)

# ----- Histograma do erro absoluto -----
plt.subplot(1, 2, 2)
plt.hist(erro_absoluto, bins=10, color='green', edgecolor='black')
plt.xlabel('Erro absoluto |software - hardware|')
plt.ylabel('Frequência')
plt.title('Histograma do Erro Absoluto')
plt.grid(True)

plt.tight_layout()
plt.show()
```
