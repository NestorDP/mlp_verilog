```python
import numpy as np
import matplotlib.pyplot as plt

# Define the sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Points at which to approximate the sigmoid
points = [-6.0, -4.0, -3.0, -2.5, -2.0, -1.3, 1.3, 2.0, 2.5, 3.0, 4.0, 6.0]

# Calculate sigmoid values at the given points
sigmoid_values = [sigmoid(x) for x in points]

# Create a fine grid for plotting the actual sigmoid curve
x_fine = np.linspace(-7, 7, 50)
y_fine = sigmoid(x_fine)

# Plot the actual sigmoid function
plt.plot(x_fine, y_fine, label="Sigmoide", color='blue')

# Plot the piece-wise linear approximation
plt.plot(points, sigmoid_values, label="Aproximação", color='red', linestyle='--', marker='o')

# Add labels and title
plt.title("Aproximação linear por partes da função sigmoide")
# Plot the sigmoid function
plt.xlabel("x")
plt.ylabel("sigmoid(x)")

# Show legend
plt.legend()

# Show the plot
plt.grid(True)
plt.show()
```

```python
# Print points and sigmoid values side by side
for point, value in zip(points, sigmoid_values):
    print(f'{point:.2f} => {value:.8f}')
```
